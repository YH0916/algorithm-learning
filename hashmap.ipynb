哈希表的底层是一个数组，他先把key通过一个hash函数，转化成数组里的索引，然后其他操作类似
hash函数的作用就是把任意长度输入的key转化成固定长度的输出（索引）index

出现hash冲突怎么办？
一种是拉链法，一种是线性探查法
拉链法，相当于存储一个链表，当多个不同的key映射到同一个索引傻姑娘，这些key->value对就存储在这个链表中
线性法，相当于发现index被占了，那就index+1，+2等等，直到找到一个空的

python中标准库提供的hash表dict， dict的键的遍历顺序就是 键的插入顺序

```python
d=dict()
d['a']=1
d['b']=2
d['c']=3
print(list(d.values()))
d['x']=4
d['y']=5
print(list(d.values()))
```

# 使用双链表实现哈希表

```python
class MyLinkedHashmap:
    class Node:
        def __init__(self, key, value):
            self.key=key
            self.value=value
            self.next=None
            self.prev=None
    
    def __init__(self):
        self.head = self.Node(None, None)
        self.tail = self.Node(None, None)
        self.head.next=self.tail
        self.tail.prev=self.head
        self.map=dict()

    def get_value(self, key):
        if key not in self.map:
            return None
        return self.map[key].val
    
    def put(self, key, val):
        # 若为新插入的节点，则同时插入链表和map
        if key not in self.map:
            # 插入新的Node
            node=self.Node(key,val)
            self.add_last_node(node)
            self.map[key]=node
            return
        # 若存在key，则替换value
        self.map[key].val=val
    
    def remove(self,key):
        if key not in self.map:
            return
        node=self.map[key]
        del self.map[key]
        self.remove_node(node)
    
    def contains_key(self, key):
        return key in self.map
    
    def keys(self):
        key_list=[]
        p=self.head.next
        while p!=self.tail:
            key_list.append(p.key)
            p=p.next
        return key_list
    
    def add_last_node(self, x):
        temp = self.tail.prev
        x.next=self.tail
        x.prev=temp

        temp.next=x
        self.tail.prev=x
    
    def remove_node(self,x):
        prev = x.prev
        next = x.next
        x.prev.next=next
        x.next.prev=prev
        x.next = x.prev =None
    
```

```python
if __name__ == '__main__':
    map = MyLinkedHashmap()
    map.put("a", 1)
    map.put("b", 2)
    map.put("c", 3)
    map.put("d", 4)
    map.put("e", 5)

    print(map.keys())  # ['a', 'b', 'c', 'd', 'e']
    map.remove("c")
    print(map.keys())  # ['a', 'b', 'd', 'e']
```

# 使用数组加强hashmap

```python
import random

class Node:
    def __init__(self, key, value):
        self.key=key
        self.value=value

class MyArrayHashMap:
    def __init__(self):
        # 存储key和key再arr中的索引
        self.map={}
        # 真正存储key-value的数组
        self.arr=[]
    
    def get(self, key):
        if key not in self.map:
            return None
        index = self.map[key]
        return self.arr[index].val
    
    def put(self, key, value):
        if self.containsKey(key):
            i=self.map[key]
            self.arr[i].val= value
            return
        self.arr.append(Node(key, value))
        self.map[key]=len(self.arr)-1
    
    def remove(self, key):
        if key not in self.map:
            return
        index=self.map[key]
        node=self.arr[index]

        e=self.arr[-1]
        self.arr[index]=e
        self.arr[-1]=node
        
        # 仔细看
        self.map[e.key]=index
        self.arr.pop()
        self.map.pop(node.key)
    
    # 随机弹出一个键
    def randomKey(self):
        n=len(self.arr)
        randomIndex=random.randint(0,n-1)
        return self.arr[randomIndex].key
    
    def containsKeys(self, key):
        return key in self.map
    
    def size(self):
        return len(self.map)

    
```
