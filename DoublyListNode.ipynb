```python
class DoublyNode:
    def __init__(self ,x):
        self.val=x
        self.prev=None
        self.next=None

def createDoublyLinkedList(arr: 'List[int]'):
    if not arr:
        return None
    
    head = DoublyNode(arr[0])
    cur = head
    for i in range(1, len(arr)):
        newnode=DoublyNode(arr[i])
        cur.next=newnode
        newnode.prev=cur
        cur=cur.next
    
    return head
```

```python
head=createDoublyLinkedList([0,1,2,3,4])
tail=None

p=head
while p:
    print(p.val)
    tail=p
    p=p.next

p=tail
while p:
    print(p.val)
    p=p.prev
```
