# 递归遍历 DFS

```python
class Node:
    def __init__(self, val:int):
        self.val=val
        self.children = [] 
    def traveres(self,root):
        if root is None:
            return
        print(root.val)#对应前序遍历
        for child in root.children:
            self.traveres(child)
        print(root.val)#对应后序遍历
```

# 层序遍历 BFS

```python
from collections import deque
class Node:
    def __init__(self, val:int):
        self.val=val
        self.children = [] 
    def traveres1(self,root):
        if root is None:
            return
        q=deque()
        q.append(root)
        while q:
            cur=q.popleft()
            print(cur.val)
            for child in cur.children:
                q.append(child)
    def traveres2(self, root):
        if root is None:
            return
        q=deque()
        q.append(root)
        depth=1
        while q:
            sz=len(q)
            for _ in range(sz):
                cur=q.popleft()
                print(cur.val)
                for child in cur.children:
                    q.append(child)   

            depth+=1 
```

```python
class State:
    def __init__(self, node, depth):
        self.node=node
        self.depth=depth

def traveres3(root):
    if root is None:
        return None
    q=deque()
    q.append(State(root,1))

    while q:
        state=q.popleft()
        cur=state.node
        depth=state.depth
        for child in cur.children:
            q.append(State(child,depth+1))

```
