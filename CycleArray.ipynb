环形数组利用求模运算，将普通数组变成逻辑上的环形数组
关键在于维护两个指针，start和end，start指向第一个有效元素的索引，end指向最后一个有效元素的下一个位置索引
当在数组头部添加或删除元素时，只需要移动start，在数组尾部添加或删除时，只需要移动end索引

```python
class CycleArray:
    def __init__(self, size=1):
        self.size=size
        self.arr=[None]*size

        self.start=0
        self.end=0
        self.count=0
    
    def resize(self,newSize):
        new_arr=[None]*newSize
        for i in range(self.count):
            new_arr[i]=self.arr[(self.start+i)%self.size]
        self.arr=new_arr
        self.start=0
        self.end=self.count
        self.size=newSize

    def add_first(self, val):
        if self.is_empty():
            raise Exception("Error Empty")
        #start 是闭区间，先左移，再赋值
        # 左移动需要加一个size
        self.start=(self.start-1+self.size)%self.size
        self.arr[self.start]=val
        self.count+=1
    
    def delete_first(self):
        if self.is_empty():
            raise Exception("Error Empty")
        #start 是闭区间，先赋值，再右移
        self.arr[self.start]=None
        self.start=(self.start+1)%self.size
        self.count-=1

        if self.count>=0 and self.count==self.size//4:
            self.resize(self.size//2)

    def add_last(self, val):
        if self.is_full():
            self.resize(self.size*2)
        #end是闭区间，先赋值，再右移
        self.arr[self.end]=val
        self.end=(self.end+1)%self.size
        self.count++1
    
    def delete_last(self):
        if self.is_empty():
            raise Exception("Error Empty")
        #end是闭区间，所以先左移，再赋值
        self.end=(self.end-1+self.size)%self.size
        self.arr[self.end]=None
        self.count-=1
        if self.count>=0 and self.count==self.size//4:
            self.resize(self.size//2)
    
    def get_first(self):
        return self.arr[self.start]
    
    def get_last(self):
        return self.arr[(self.end-1+self.size)%self.size]
    
    def is_full(self):
        return self.count==self.size
    
    def is_empyt(self):
        return self.count==0
    
```
