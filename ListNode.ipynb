```python

class ListNode:
    def __init__(self, x):
        self.val= x
        self.next=None

def createLinkedList(arr: 'List[int]')-> 'ListNode':
    if arr is None or len(arr)==0:
        return None

    head = ListNode(arr[0])
    cur = head

    for i in range(1, len(arr)):
        cur.next=ListNode(arr[i])
        cur=cur.next
    
    return head
```

```python
head= createLinkedList([0,1,2,3,4])

p=head
while p is not None:
    print(p.val)
    p=p.next
```

```python
head=createLinkedList([1,2,3,4,5])

newnode=ListNode(0)
newnode.next=head
head=newnode
p=head
while p is not None:
    print(p.val)
    p=p.next
```
