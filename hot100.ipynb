# 无重复字符的最长子串

````python
class Solution(object):
    def lengthOfLongestSubstring(self, s):
        """
        :type s: str
        :rtype: int
        """
        if s is None:
            return 0
        res=0
        new_char={}
        left=right=0
        length=len(s)
        while(right<length):
            char=s[right]
            new_char[char] = new_char.get(char,0)+1
            right+=1
            while new_char[char]>1:
                new_char[s[left]]=new_char.get(s[left],0)-1
                left+=1
            res=max(res, right-left)
        return res
            
````

# 滑动窗口最大值

````python
def maxSlidingWindow(nums, k):
    """
    :type nums: List[int]
    :type k: int
    :rtype: List[int]
    """
    left=0
    right=0+k
    res=[]
    currentMax=float('-inf')
    while right<=len(nums):
        for i in range(left,right):
            currentMax = max(currentMax,nums[i])
        res.append(currentMax)
        currentMax=float('-inf')
        left+=1
        right+=1
    return res
````

# 相交链表

````python
class Solution(object):
    def getIntersectionNode(self, headA, headB):
        """
        :type head1, head1: ListNode
        :rtype: ListNode
        """
        p1=headA
        p2=headB
        while p1!=p2:
            p1=headB if p1 is None else p1.next
            p2=headA if p2 is None else p2.next
        return p1
    
````

# 反转链表 重点回顾

````python
class Solution(object):
    def reverseList(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        if head is None:
            return
        cur = head
        prev=None
        while cur is not None:
            newnode = cur.next
            cur.next=prev
            prev=cur
            cur=newnode
        return prev
        
````

# 回文链表

````python
class Solution(object):
    def isPalindrome(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: bool
        """
        # 快慢指针找到中点，
        slow=fast=head
        while fast and fast.next:
            slow=slow.next
            fast=fast.next.next
        # 然后反转后半部分列表
        second_half=self.reverseList(slow)
        p1=head
        p2=second_half
        # 然后比较
        while p1 and p2:
            if p1.val != p2.val:
                return False
            p1=p1.next
            p2=p2.next
        return True 

    def reverseList(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        if head is None:
            return
        cur = head
        prev=None
        while cur is not None:
            newnode = cur.next
            cur.next=prev
            prev=cur
            cur=newnode
        return prev
````

# 环形链表

````python
class Solution(object):
    def hasCycle(self, head):
        """
        :type head: ListNode
        :rtype: bool
        """
        slow=fast=head
        while fast and fast.next:
            slow=slow.next
            fast=fast.next.next
            if slow == fast:
                return True
            
        return False
````

# 合并两个有序链表

````python
# Definition for singly-linked list.
# class ListNode(object):
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution(object):
    def mergeTwoLists(self, list1, list2):
        """
        :type list1: Optional[ListNode]
        :type list2: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        dummy = ListNode(-1)
        p=dummy
        p1=list1
        p2=list2
        while p1 and p2:
            if p1.val > p2.val:
                p.next = p2
                p2=p2.next
            else:
                p.next=p1
                p1=p1.next
            p=p.next
        
        if p1: p.next=p1
        if p2: p.next=p2

        return dummy.next #前面的节点本来是虚拟头节点，现在他的next才是我们需要的链表
````

# 单链表的分解

可以把原来的链表分成两个小链表，一个链表中的元素大小都小于x，另一个中的都大于x，然后接起来就行

````python
# Definition for singly-linked list.
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution(object):
    def partition(self, head, x):
        """
        :type head: Optional[ListNode]
        :type x: int
        :rtype: Optional[ListNode]
        """
        dummy1=ListNode(-1)
        dummy2=ListNode(-2)
        p1, p2 = dummy1, dummy2
        p=head
        while p:
            if p.val>= x:
                p2.next=p
                p2=p2.next
            else:
                p1.next=p 
                p1=p1.next
            # 断开原来链表中的每个节点的next指针（如果出现环，那么他可能就会一直循环），需要注意这是个好习惯！！！！
            temp = p.next
            p.next=None
            p=temp
        
        p1.next=dummy2.next
        return dummy1.next
````

# 合并K个升序链表

把链表节点放入一个最小堆，然后每次可以获得k个节点中的最小节点

设所有链表总共有 N 个节点、链表数为 K：
每个节点入堆一次、出堆一次，单次 O(log K)，总计 O(N log K)。

这是合并 K 路有序序列的最优典型做法。

````python
import heapq
class Solution(object):
    def contrast(self, node1, node2):
        return node1.val < node2.val
    
    def mergeKLists(self, lists):
        """
        :type lists: List[Optional[ListNode]]
        :rtype: Optional[ListNode]
        """
        if not lists:
            return None
        dummy1=ListNode(-1)
        p=dummy1

        # 构造优先级队列
        pq=[]
        # 将K个链表的头节点依次加入最小堆
        for i, head in enumerate(lists):
            if head:
                # heappush 是向堆中加入新的元素
                # heapd 会按照元组的字典序比较，先比较第一项，依次比较后面的
                # head.val：作为优先级键，保证堆顶总是当前最小值。
                # i：打破平局（tie-breaker）。当两个节点值相同，比较 i 就能分出大小，
                # 避免 Python 去比较第三项（链表节点对象），否则在 Python3 里会因为对象不可比而 TypeError。
                # head：把节点本体带上，弹出时好接到结果链表里。
                heapq.heappush(pq, (head.val, i, head))
        while pq:
            # 然后依次获取堆顶部的元素，加入到一个新的队列就行
            val, i, node =heapq.heappop(pq)
            p.next=node
            if node.next:
                heapq.heappush(pq, (node.next.val, i, node.next))
            p=p.next
        return dummy1.next

````

## headp库函数的使用

````python
import heapq

# 初始化列表并转为最小堆
nums = [5, 7, 3, 1, 9]
heapq.heapify(nums)
print(nums)  # [1, 5, 3, 7, 9] （最小堆结构）

# 插入元素
heapq.heappush(nums, 2)
print(nums)  # [1, 2, 3, 7, 9, 5]

# 弹出最小元素
min_val = heapq.heappop(nums)
print(min_val)  # 1
print(nums)     # [2, 5, 3, 7, 9]

# 获取最小两个元素（不破坏堆）
print(heapq.nsmallest(2, nums))  # [2, 3]

# 获取最大两个元素（不破坏堆）
print(heapq.nlargest(2, nums))   # [9, 7]
````

## 删除单链表中的倒数第K个节点

````python
class Solution(object):
    def removeNthFromEnd(self, head, n):
        """
        :type head: Optional[ListNode]
        :type n: int
        :rtype: Optional[ListNode]
        """
        dummy= ListNode(-1)
        dummy.next=head
        p1=dummy.next
        for _ in range(n):
            p1=p1.next
        p2=dummy.next
        p=dummy
        while p1!=None:
            p1=p1.next
            p2=p2.next
            p=p.next
        temp=p2.next
        p2.next=None
        p.next=temp
        return dummy.next
````

# 零钱兑换

### 暴力递归

````python
class Solution(object):
    def coinChange(self, coins, amount):
        return self.dp(coins, amount)

    # 定义一个dp函数，要凑出金额n，至少要dp(coins, n)个硬币
    def dp(self, coins, n):
        if n==0:return 0
        if n<0: return -1
        res=float('inf')
        for coin in coins:
            SubProblem = self.dp(coins, n-coin)
            if SubProblem ==-1:
                continue
            res=min(res, SubProblem+1)
        
        return res if res!=float('inf') else -1
````

### 自顶向下解法  带备忘录

````python
class Solution(object):
    count=[]
    def coinChange(self, coins, amount):
        self.count=[-100]*(amount+1)
        return self.dp(coins, amount)

    # 定义一个dp函数，要凑出金额n，至少要dp(coins, n)个硬币
    def dp(self, coins, amount):
        if amount==0:return 0
        if amount<0: return -1
        if self.count[amount]!=-100:
            return self.count[amount]
        res = float('inf')
        for coin in coins:
            SubProblem = self.dp(coins, amount-coin)
            if SubProblem ==-1:
                continue
            res=min(res, SubProblem+1)
        self.count[amount] = res if res!=float('inf') else -1
        return self.count[amount]
````

### dp数组的循环解法

````python
class Solution(object):
    def coinChange(self, coins, amount):
        dp=[amount+1]*(amount+1)# 注意关注初始化，所有可能的总钱数里 最多就是amount个硬币
        dp[0]=0
        for i in range(len(dp)):
            for coin in coins:
                if i-coin<0:
                    continue
                dp[i]=min(dp[i], dp[i-coin]+1)        
        return -1 if dp[amount]==amount+1 else dp[amount]
    
````

### 暴力递归

````python
class Solution(object):
    def rob(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        return self.dp(nums, 0)
    # 用于返回从nums[start...]开始抢的最大金额
    def dp(self, nums, start):
        if start >= len(nums):
            return 0
        res=max(
            self.dp(nums, start+1),
            self.dp(nums, start+2)+nums[start])
        return res
````

````python
class Solution(object):
    def rob(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        self.temp=[-1]*len(nums)
        return self.dp(nums, 0)
    # 用于返回从nums[start...]开始抢的最大金额
    def dp(self, nums, start):
        if start >= len(nums):
            return 0
        if self.temp[start]!=-1:
            return self.temp[start]
        res=max(
            self.dp(nums, start+1),
            self.dp(nums, start+2)+nums[start])
        self.temp[start]=res
        return res
````

### 自底向上

````python
#从前往后算
class Solution:
    def rob(self, nums):
        n=len(nums)
        if n==0: return 0
        if n==1: return nums[0]
        dp_i_2=nums[0]
        dp_i_1=max(nums[0],nums[1])
        for i in range(2, n):
            dp_i=max(dp_i_1, dp_i_2+nums[i])
            dp_i_2,dp_i_1 = dp_i_1,dp_i
        return dp_i
    
````

````python
#从后往前算
class Solution:
    # dp[i]=从第 i 间房子到最后一间，能偷到的最大金额
    def rob(self, nums):
        n =len(nums)
        dp_i_1,dp_i_2=0,0
        dp_i = 0
        for i in range(n-1, -1,-1):
        # 如果你从前往后算，dp[i+1] 和 dp[i+2] 还没计算出来
        # 所以要 从后往前算，保证依赖的值已经被算出来了
            dp_i=max(dp_i_1, nums[i]+dp_i_2)
            dp_i_2=dp_i_1
            dp_i_1=dp_i
        return dp_i
````

````python
class Solution(object):
    def rob(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        length=len(nums)
        if length==0:return 0
        if length==1:return nums[0]
        return max(self.robbery(nums,0,length-2),self.robbery(nums,1,length-1))
    # 用于返回从nums[start, end]抢的最大金额
    def robbery(self, nums, start, end):
        dp_i_1,dp_i_2=0,0
        dp_i=0
        for i in range(end, start-1, -1):
            dp_i = max(dp_i_1, dp_i_2+nums[i])
            dp_i_2, dp_i_1 =dp_i_1, dp_i
        return dp_i
````

# 单词拆分

### 动态规划

````python
class Solution(object):
    def wordBreak(self, s, wordDict):
        """
        :type s: str
        :type wordDict: List[str]
        :rtype: bool
        """
        hashword=set(wordDict)
        n=len(s)+1
        dp=[False]*(n+1)
        dp[0]=True
        # 外层循环：枚举字符串的前缀长度 i
        for i in range(1, n+1):
            # 内层循环：枚举拆分点 j，把 s[0:i] 分成 s[0:j] 和 s[j:i]
            for j in range(i):
                if dp[j] and s[j:i] in hashword:
                    # 条件1: s[0:j] 已经能被拆分（dp[j] == True）
                    # 条件2: s[j:i] 是字典里的单词
                    dp[i]=True
                    break
        return dp[n]# 注意这里写的是 dp[n]，实际上应返回 dp[len(s)]
````

### 自顶向下 带备忘录解法

````python
class Solution(object):
    def wordBreak(self, s, wordDict):
        self.hashword=set(wordDict)
        self.temp={}
        return self.dp(s,0)
    # 函数的主要意思是，从s[start：]开始，其能否被拆分为字典中的单词
    def dp(self, s, start):
        if start==len(s):return True 
        #如果start是代表字符串结束，
        #意味着已经遍历完s，把他都拆干净了，
        #证明他可以被拆分
        if start in self.temp:
            return self.temp[start]
        for end in range(start+1, len(s)+1):
            if s[start:end] in self.hashword and self.dp(s, end):
                self.temp[start]=True
                return True
        self.temp[start]=False
        return False
````

# 最长递增子序列

给你一个整数数组 nums ，找到其中最长严格递增子序列的长度。
子序列 是由数组派生而来的序列，删除（或不删除）数组中的元素而不改变其余元素的顺序。例如，[3,6,2,7] 是数组 [0,3,1,6,2,2,7] 的子序列。
示例 1：
输入：nums = [10,9,2,5,3,7,101,18]
输出：4
解释：最长递增子序列是 [2,3,7,101]，因此长度为 4 。
示例 2：
输入：nums = [0,1,0,3,2,3]
输出：4
示例 3：
输入：nums = [7,7,7,7,7,7,7]
输出：1

````python
class Solution(object):
    def lengthOfLIS(self, nums):
        if nums is None:
            return 0
        dp=[1]*len(nums)
        for i in range(len(nums)):
            for j in range(i):
                if nums[i]>nums[j]:
                    dp[i]=max(dp[i],dp[j]+1)
        return max(dp)
````

# 乘积最大子数组

精彩啊

给你一个整数数组 nums ，请你找出数组中乘积最大的非空连续 子数组（该子数组中至少包含一个数字），并返回该子数组所对应的乘积。
测试用例的答案是一个 32-位 整数。
示例 1:
输入: nums = [2,3,-2,4]
输出: 6
解释: 子数组 [2,3] 有最大乘积 6。
示例 2:
输入: nums = [-2,0,-1]
输出: 0
解释: 结果不能为 2, 因为 [-2,-1] 不是子数组。

````python
class Solution(object):
    def maxProduct(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        if nums is None:
            return 0
        n=len(nums)
        dp_max=[0]*n
        dp_min=[0]*n
        dp_max[0]=dp_min[0]=nums[0]
        res=nums[0]
        for i in range(1,n):
            dp_max[i]=max(nums[i], nums[i]*dp_max[i-1], nums[i]*dp_min[i-1])
            dp_min[i]=min(nums[i], nums[i]*dp_max[i-1], nums[i]*dp_min[i-1])
            res=max(res, dp_max[i])
        return res
````

### 递归+回溯

相当于是 一个树，选还是不选的问题

'''
                          sum=0
                        /       \
               不选1(sum=0)     选1(sum=1)
                 /      \         /       \
        不选5(sum=0)  选5(sum=5)  不选5(sum=1)  选5(sum=6)
           ...           ...          ...         ...
'''

````python
class Solution(object):
    def canPartition(self, nums):
        total = sum (nums)
        if total % 2 ==1:
            return False
        total = nums // 2
        # 当前考虑第i个数字，current_num为当前已经选择的数字的总和
        def dp(i, current_num):
            if current_num==total:
                return True
            if current_num>total or i==len(nums):
                return False
            return dp(i+1, current_num+nums[i]) or dp(i+1, current_num)
        
        return dp(0,0)
        
````

### 带备忘录解法

````python
class Solution(object):
    def canPartition(self, nums):
        total = sum (nums)
        if total % 2 ==1:
            return False
        total = total // 2
        temp={}
        # 当前考虑第i个数字，current_num为当前已经选择的数字的总和
        def dp(i, current_num):
            if current_num==total:
                return True
            if (i, current_num) in temp:
                return temp[(i, current_num)]
            remain=sum(nums[i:])
            if current_num>total or current_num+remain<total or i==len(nums):
                return False
            res = dp(i+1, current_num+nums[i]) or dp(i+1, current_num)
            temp[(i, current_num)]=res
            return res
        
        return dp(0,0)
        
````

（简化思路） 0-1背包问题，拿或者不拿，只能拿一次

dp[i]=dp[i] or dp[i−weight]

需要倒序遍历，为什么？
更新 dp[i] 时，使用 dp[i - weight] 来表示 之前轮的状态

如果正序，会使用当前轮刚更新的 dp[i-weight] → 当前物品被重复使用 → 不符合 0-1 背包

✅ 所以 0-1 背包必须 倒序遍历

````python
class Solution(object):
    def canPartition(self, nums):
        total = sum(nums)
        # 如果总和是奇数，不可能分割
        if total % 2 != 0:
            return False
        target = total // 2
        dp = [False] * (target+1)
        dp[0]=True # 和为0时总是True
        for num in nums:
            for i in range(target, num-1, -1):
                dp[i]=dp[i] or dp[i - num]
        return dp[target]
````

# 不同路径

一个机器人位于一个 m x n 网格的左上角 （起始点在下图中标记为 “Start” ）。

机器人每次只能向下或者向右移动一步。机器人试图达到网格的右下角（在下图中标记为 “Finish” ）。

问总共有多少条不同的路径？

````python
class Solution(object):
    def uniquePaths(self, m, n):
        dp=[[0]*(n) for _ in range(m)]
        for j in range(n):
            dp[0][j]=1
        for i in range(m):
            dp[i][0]=1
        for i in range(1, m):
            for j in range(1,n):
                dp[i][j]=dp[i-1][j]+dp[i][j-1]
        return dp[m-1][n-1]
````

# 最小路径和
给定一个包含非负整数的 m x n 网格 grid ，请找出一条从左上角到右下角的路径，使得路径上的数字总和为最小。
说明：每次只能向下或者向右移动一步。

````python
class Solution(object):
    def minPathSum(self, grid):
        """
        :type grid: List[List[int]]
        :rtype: int
        """
        m,n = len(grid), len(grid[0])
        for i in range(1,m):
            grid[i][0]+=grid[i-1][0]
        for j in range(1,n):
            grid[0][j]+=grid[0][j-1]
        for i in range(1, m):
            for j in range(1, n):
                grid[i][j]=min(grid[i-1][j], grid[i][j-1])+grid[i][j]
        return grid[m-1][n-1]
        
````

# 最长回文子串
给你一个字符串 s，找到 s 中最长的 回文 子串。

因为考虑到dp[i][j]依赖于dp[i-1][j+1],所以必须是从下往上填

````python
class Solution(object):
    def longestPalindrome(self, s):
        """
        :type s: str
        :rtype: str
        """
        n=len(s)
        if n<2:
            return s
        max_len=1
        start=0
        dp=[[False]*(n) for _ in range(n)]
        for i in range(n): dp[i][i]=True
        for i in range(1,n):
            for j in range(i):
                if s[j]==s[i]:
                    if i-j<3:
                        dp[j][i]=True
                    else:
                        dp[j][i]=dp[j+1][i-1]
                if dp[j][i] and i-j+1>max_len:
                    max_len=i-j+1
                    start=j 
        
        return s[start:start+max_len]
````

暴力解法

````python
class Solution(object):
    def longestPalindrome(self, s):
        """
        :type s: str
        :rtype: str
        """
        def isParlindrome(s):
            left, right =0, len(s)-1
            while left<right:
                if s[left]!=s[right]:
                    return False
                left+=1
                right-=1
            return True
        n=len(s)
        if n<2:
            return s
        max_len=1
        start=0
        for i in range(n):
            for j in range(i+1,n):
                if isParlindrome(s[i:j+1]):
                    cur_len=len(s[i:j+1])
                    if cur_len>max_len:
                        max_len=cur_len
                        start=i
        return s[start:start+max_len]
````

# 最长公共子序列
给定两个字符串 text1 和 text2，返回这两个字符串的最长 公共子序列 的长度。如果不存在 公共子序列 ，返回 0 。

一个字符串的 子序列 是指这样一个新的字符串：它是由原字符串在不改变字符的相对顺序的情况下删除某些字符（也可以不删除任何字符）后组成的新字符串。

例如，"ace" 是 "abcde" 的子序列，但 "aec" 不是 "abcde" 的子序列。
两个字符串的 公共子序列 是这两个字符串所共同拥有的子序列。

{贪心法，不能获得正确结果}

````python
class Solution(object):
    def longestCommonSubsequence(self, text1, text2):
        """
        :type text1: str
        :type text2: str
        :rtype: int
        """
        m, n=len(text1),len(text2)
        if m>n:
            text1, text2 = text2, text1
            m,n=n,m
        
        start=0
        max_len=0
        for i in range(m):
            for j in range(start, n):
                if text1[i]==text2[j]:
                    max_len+=1
                    start=j+1
                    break
        return max_len
````

标准的dp

状态转移方程如下：
```
dp[i][j]=前 i 个字符的 text1 和 前 j 个字符的 text2 的最长公共子序列长度
```
转移思路
```
如果 text1[i-1] == text2[j-1]，那么这两个字符可以加入到 LCS 中。
dp[i][j]=dp[i−1][j−1]+1
如果 text1[i-1] != text2[j-1], 那么就要移动某个字符串，在下面两种情况里取最大值：
不要 text1[i-1] → 看 dp[i-1][j]
不要 text2[j-1] → 看 dp[i][j-1]
dp[i][j]=max(dp[i−1][j],dp[i][j−1])
```

````python
class Solution(object):
    def longestCommonSubsequence(self, text1, text2):
        """
        :type text1: str
        :type text2: str
        :rtype: int
        """
        m,n =len(text1, text2)
        dp=[[0]*(n+1) for _ in range(m+1)]
        for i in range(1, m+1):
            for j in range(1, n+1):
                if text1[i-1]==text2[i-1]:
                    dp[i][j]=dp[i-1][j-1]+1
                else:
                    dp[i][j]=max(dp[i-1][j], dp[i][j-1])
        return dp[m][n]
````

# 编辑距离
```
解决两个字符串的动态规划问题，一般都是用两个指针i，j分别指向两个字符串的头部和尾部，然后尝试写状态转移方程
```
给你两个单词 word1 和 word2， 请返回将 word1 转换成 word2 所使用的最少操作数  。

你可以对一个单词进行如下三种操作：

（1）插入一个字符
（2）删除一个字符
（3）替换一个字符

base case是i走完s1或者j走完s2，直接返回另一串没走完的字符串剩下的长度

对于每一对字符串s1[1]和s2[j],有四种操作：

```
if s1[i] == s2[j]:
    啥都别做（skip）
    i, j 同时向前移动
else:
    三选一：
        插入（insert）
        删除（delete）
        替换（replace）
```

### 暴力解法

````python
class Solution(object):
    def minDistance(self, word1, word2):
        m, n = len(word1), len(word2)
        return self.dp(word1, m-1, word2, n-1)
    def dp(self, word1, i, word2, j):
        if i == -1: return j+1
        if j == -1: return i+1
        if word1[i]==word2[j]:
            return self.dp(word1, i-1, word2, j-1)
        return min(
            # 插入
            self.dp(word1, i, word2, j-1)+1,
            # 删除
            self.dp(word1, i-1, word2, j)+1,
            # 替换
            self.dp(word1, i-1, word2, j-1)+1
        )
````

### 带备忘录的递归解法（自顶向下）

````python
class Solution(object):
    def minDistance(self, word1, word2):
        m, n = len(word1), len(word2)
        self.temp=[[-1]*(n) for _ in range(m)]
        return self.dp(word1, m-1, word2, n-1)
    def dp(self, word1, i, word2, j):
        if i == -1: return j+1
        if j == -1: return i+1
        if self.temp[i][j]!=-1:return self.temp[i][j]
        if word1[i]==word2[j]:
            self.temp[i][j]=self.dp(word1, i-1, word2, j-1)
        else:
            self.temp[i][j]=min(
                # 插入
                self.dp(word1, i, word2, j-1)+1,
                # 删除
                self.dp(word1, i-1, word2, j)+1,
                # 替换
                self.dp(word1, i-1, word2, j-1)+1
            )
        return self.temp[i][j]
````
